# Rainfall Analysis of Ahilyanagar District 

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("RainfallDataset.csv")
```

```python
print(df.head())
```

```python
print(df.head())
print(df.info())
print(df.columns)
```

```python
print(df.isnull().sum())
```

```python
print(df.describe())
```

```python
print("Rows and Columns:")
print(df.shape)
```

```python
total_rainfall = df.iloc[:, 1:].sum()
print(total_rainfall)
```

```python
print("Highest Rainfall Taluka:")
print(total_rainfall.idxmax())
print("Rainfall:")
print(total_rainfall.max())
```

```python
print("Lowest Rainfall Taluka:")
print(total_rainfall.idxmin())

print("Rainfall:")
print(total_rainfall.min())
```

```python
for rain in df["DISTRICT - Ahilyanagar(Ahmadnagar)"]:
    if rain > 200:
        print("Heavy Rainfall")
    elif rain > 100:
        print("Moderate Rainfall")
    else:
        print("Low Rainfall")
```

```python
district = "DISTRICT - Ahilyanagar(Ahmadnagar)"
top5 = df.nlargest(5, district)
print(top5[["Date With Month wise Rainfall in mm", district]])
```

```python
district = "DISTRICT - Ahilyanagar(Ahmadnagar)"

print("Average Rainfall:", df[district].mean())
print("Highest Rainfall:", df[district].max())
print("Lowest Rainfall:", df[district].min())
print("Total Rainfall:", df[district].sum())
```

```python
def category(rain):
    if rain > 200:
        return "Heavy"
    elif rain > 100:
        return "Moderate"
    else:
        return "Low"
df["Rainfall_Category"] = df[district].apply(category)
print(df.head())
```

```python
rain = np.array(df["DISTRICT - Ahilyanagar(Ahmadnagar)"])
print("Mean:", np.mean(rain))
print("Maximum:", np.max(rain))
print("Minimum:", np.min(rain))
```

```python
for i in range(len(df)):
    if df["Rahata"][i] > df["Rahuri"][i]:
        print("Rahata higher in", df["Date With Month wise Rainfall in mm"][i])
    else:
        print("Rahuri higher in", df["Date With Month wise Rainfall in mm"][i])
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10,5))
plt.plot(df["Date With Month wise Rainfall in mm"],
         df["DISTRICT - Ahilyanagar(Ahmadnagar)"],
         marker='o')

plt.title("District Rainfall Trend")
plt.xlabel("Month")
plt.ylabel("Rainfall ") 
plt.show()
```
